```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
data = {
    "x": [10, 0, 5, 6, 3],
    "y": [6, 7, 4, 2, 3]
}
```

```python
df = pd.DataFrame(data)
x = df["x"].values
y = df["y"].values
```

```python
mu_a = 0.0
sigma_a = 1.0

mu_b = 0.0
sigma_b = 1.0

sigma_noise = 1.0

n_samples = 10000
```

```python
elbo_samples = []

for _ in range(n_samples):

    a = np.random.normal(mu_a, sigma_a)
    b = np.random.normal(mu_b, sigma_b)

    y_pred = a * x + b

    log_likelihood = np.sum(
        norm.logpdf(y, loc=y_pred, scale=sigma_noise)
    )

    log_prior = (
        norm.logpdf(a, 0, 1) +
        norm.logpdf(b, 0, 1)
    )

    log_q = (
        norm.logpdf(a, mu_a, sigma_a) +
        norm.logpdf(b, mu_b, sigma_b)
    )


    elbo = log_likelihood + log_prior - log_q

    elbo_samples.append(elbo)

estimated_elbo = np.mean(elbo_samples)

print("Estimated ELBO:", estimated_elbo)
```
